## Questão 1 — Criando um array

Crie um array NumPy contendo as notas `7.5, 8.0, 6.5, 9.0, 4.5, 8.5, 7.0, 5.5, 10.0, 6.0` e exiba:

## Questão 2 — Estatísticas básicas

Utilizando o array da questão anterior, determine:

a) média;  
b) mediana;  
c) maior nota;  
d) menor nota;  
e) desvio padrão;  
f) variância.

Apresente os resultados com duas casas decimais.

````python
media = np.mean(notas)
mediana = np.median(notas)
maior_nota = np.max(notas)
menor_nota = np.min(notas)
desvio_padrao = np.std(notas)
variancia = np.var(notas)

print(f"a) Média: {media:.2f}")
print(f"b) Mediana: {mediana:.2f}")
print(f"c) Maior nota: {maior_nota:.2f}")
print(f"d) Menor nota: {menor_nota:.2f}")
print(f"e) Desvio padrão: {desvio_padrao:.2f}")
print(f"f) Variância: {variancia:.2f}")
````

## Questão 4 — Operações vetorizadas

Considere que o array abaixo contém temperaturas em graus Celsius:

`temperaturas = np.array([25, 27, 29, 31, 32, 30, 28, 26])`

Converta todas as temperaturas para Fahrenheit utilizando uma única operação vetorizada:

$$F = C \times \frac{9}{5} + 32$$

Não utilize `for`.

Depois determine as temperaturas média, mínima e máxima nas duas escalas.

````python
temperaturas = np.array([25, 27, 29, 31, 32, 30, 28, 26])

# Conversão vetorizada: a fórmula é aplicada a todos os elementos de uma vez
temperaturas_f = temperaturas * 9 / 5 + 32

print("Celsius:   ", temperaturas)
print("Fahrenheit:", temperaturas_f)
print()
print(f"Média:  {np.mean(temperaturas):.2f} °C | {np.mean(temperaturas_f):.2f} °F")
print(f"Mínima: {np.min(temperaturas):.2f} °C | {np.min(temperaturas_f):.2f} °F")
print(f"Máxima: {np.max(temperaturas):.2f} °C | {np.max(temperaturas_f):.2f} °F")
````

**Observações:**

- A expressão `temperaturas * 9 / 5 + 32` é aplicada elemento a elemento pelo NumPy, sem necessidade de `for`. Isso é o que se chama de **operação vetorizada**: o código fica mais curto e é executado muito mais rápido que um laço em Python.
- O array original é de inteiros (`int64`), mas o resultado passa a ser `float64`, pois a divisão `/` sempre produz números de ponto flutuante.
- Como a conversão é uma transformação linear crescente, a média, a mínima e a máxima em Fahrenheit correspondem exatamente à conversão dos valores em Celsius. Por exemplo, a média 28,5 °C × 9/5 + 32 = 83,3 °F.

## Questão 5 — Máscaras booleanas

Considere:

`idades = np.array([17, 18, 21, 16, 25, 34, 15, 22, 40, 19])`

Utilizando máscaras booleanas:

a) exiba somente as idades maiores ou iguais a 18;  
b) conte quantas pessoas são menores de idade;  
c) exiba somente as idades entre 18 e 25 anos, inclusive;  
d) calcule a idade média apenas das pessoas maiores de idade.

````python
idades = np.array([17, 18, 21, 16, 25, 34, 15, 22, 40, 19])

maiores_de_idade = idades >= 18
menores_de_idade = idades < 18
entre_18_e_25 = (idades >= 18) & (idades <= 25)

print("Máscara (idades >= 18):", maiores_de_idade)
print()
print("a) Idades >= 18:", idades[maiores_de_idade])
print("b) Quantidade de menores de idade:", np.sum(menores_de_idade))
print("c) Idades entre 18 e 25:", idades[entre_18_e_25])
print(f"d) Idade média dos maiores de idade: {np.mean(idades[maiores_de_idade]):.2f}")
````

**Observações:**

- Uma comparação como `idades >= 18` não devolve um único valor, e sim um **array de `True`/`False`** do mesmo tamanho, chamado de máscara booleana.
- Ao usar a máscara como índice (`idades[mascara]`), o NumPy retorna apenas os elementos nas posições marcadas como `True`.
- Para contar, usa-se `np.sum()` na máscara: `True` vale 1 e `False` vale 0, então a soma é o número de elementos que atendem à condição (3 menores de idade: 17, 16 e 15).
- Para combinar condições, usa-se `&` (e) ou `|` (ou), com **parênteses** em cada condição. As palavras `and`/`or` do Python não funcionam com arrays.
- A média dos maiores de idade (25,57) considera apenas os 7 valores selecionados pela máscara, e não o array inteiro.

```python
vendas = np.array([
```

Determine:

b) média mensal;  
c) mês de maior venda;

e) quantidade de meses com vendas superiores à média.

````python
vendas = np.array([
    12500, 13800, 11900, 14500,
    15200, 16000, 15800, 17100,
    16900, 18200, 19500, 21000
])

nomes_meses = np.array(["Janeiro", "Fevereiro", "Março", "Abril", "Maio", "Junho",
                        "Julho", "Agosto", "Setembro", "Outubro", "Novembro", "Dezembro"])

venda_total = np.sum(vendas)
media_mensal = np.mean(vendas)
indice_maior = np.argmax(vendas)
indice_menor = np.argmin(vendas)
acima_da_media = vendas > media_mensal

print(f"a) Venda total anual: R$ {venda_total:,.2f}")
print(f"b) Média mensal: R$ {media_mensal:,.2f}")
print(f"c) Mês de maior venda: {nomes_meses[indice_maior]} (mês {indice_maior + 1}) - R$ {vendas[indice_maior]:,.2f}")
print(f"d) Mês de menor venda: {nomes_meses[indice_menor]} (mês {indice_menor + 1}) - R$ {vendas[indice_menor]:,.2f}")
print(f"e) Meses com vendas acima da média: {np.sum(acima_da_media)} -> {nomes_meses[acima_da_media]}")
````

**Observações:**

- No f-string, `:,.2f` adiciona separador de milhar e duas casas decimais.

Utilize os dados da questão anterior.

Calcule a variação absoluta das vendas entre cada par de meses consecutivos.

Exemplo:


Em seguida identifique:

a) maior crescimento;

## Questão 8 — Notas de uma turma

Cada linha representa um aluno e cada coluna uma avaliação:

```python
notas = np.array([
    [7.0, 8.0, 9.0],
    [5.0, 6.0, 7.0],
    [8.5, 9.0, 9.5],
    [4.0, 5.0, 6.0],
    [6.5, 7.5, 8.0],
    [9.0, 8.5, 10.0]
])
```

Calcule:

a) média de cada aluno;  
b) média de cada avaliação;  
c) aluno com maior média;  
d) menor nota da turma;  
e) quantidade de alunos com média ≥ 7.

````python
# Nome diferente de "notas" para não sobrescrever o array da Questão 1
notas_turma = np.array([
    [7.0, 8.0, 9.0],
    [5.0, 6.0, 7.0],
    [8.5, 9.0, 9.5],
    [4.0, 5.0, 6.0],
    [6.5, 7.5, 8.0],
    [9.0, 8.5, 10.0]
])

print("Shape (alunos, avaliações):", notas_turma.shape)
print()

media_alunos = np.mean(notas_turma, axis=1)       # axis=1 -> percorre as colunas de cada linha
media_avaliacoes = np.mean(notas_turma, axis=0)   # axis=0 -> percorre as linhas de cada coluna
indice_melhor_aluno = np.argmax(media_alunos)
menor_nota_turma = np.min(notas_turma)
aluno_menor, avaliacao_menor = np.unravel_index(np.argmin(notas_turma), notas_turma.shape)
aprovados = media_alunos >= 7

print("a) Média de cada aluno:", np.round(media_alunos, 2))
print("b) Média de cada avaliação:", np.round(media_avaliacoes, 2))
print(f"c) Aluno com maior média: Aluno {indice_melhor_aluno + 1} (média {media_alunos[indice_melhor_aluno]:.2f})")
print(f"d) Menor nota da turma: {menor_nota_turma:.2f} (Aluno {aluno_menor + 1}, Avaliação {avaliacao_menor + 1})")
print(f"e) Alunos com média >= 7: {np.sum(aprovados)} -> Alunos {np.where(aprovados)[0] + 1}")
````

**Observações:**

- A matriz tem `shape (6, 3)`: 6 linhas (alunos) e 3 colunas (avaliações).
- O parâmetro `axis` define a direção do cálculo:
  - `axis=1` calcula ao longo das colunas de cada linha, gerando **uma média por aluno** (6 valores);
  - `axis=0` calcula ao longo das linhas de cada coluna, gerando **uma média por avaliação** (3 valores).
- Sem `axis`, `np.min()` considera a matriz inteira e retorna a menor nota da turma (4,0).
- `np.argmin()` em uma matriz retorna a posição como se ela fosse um vetor achatado. `np.unravel_index()` converte essa posição em (linha, coluna), permitindo saber qual aluno e qual avaliação.
- A média das avaliações cresce da 1ª (6,67) para a 3ª (8,25): a turma melhorou ao longo do período, e todos os alunos tiveram a 3ª nota maior que a 1ª.
- 4 dos 6 alunos (1, 3, 5 e 6) têm média ≥ 7. Os alunos 2 e 4 ficaram abaixo, com médias 6,0 e 5,0.

## Questão 10 — Reajuste de preços

Considere:

`precos = np.array([10.50, 20.00, 15.75, 32.90, 8.99, 45.50])`

Aplique:

- reajuste de 8% para todos os produtos;
- posteriormente desconto de 5% nos produtos cujo novo preço seja superior a R$ 30,00.

Apresente os preços originais e finais.

A solução deve utilizar operações vetorizadas.

````python
precos = np.array([10.50, 20.00, 15.75, 32.90, 8.99, 45.50])

# 1) Reajuste de 8% em todos os produtos
precos_reajustados = precos * 1.08

# 2) Desconto de 5% apenas onde o novo preço passou de R$ 30,00
recebe_desconto = precos_reajustados > 30
precos_finais = np.where(recebe_desconto, precos_reajustados * 0.95, precos_reajustados)

print("Preços originais:   ", precos)
print("Após reajuste de 8%:", np.round(precos_reajustados, 2))
print("Recebe desconto?    ", recebe_desconto)
print("Preços finais:      ", np.round(precos_finais, 2))
````

**Observações:**

- Reajuste de 8% equivale a multiplicar por **1,08**, e desconto de 5% equivale a multiplicar por **0,95**. Cada multiplicação é aplicada ao array inteiro de uma vez.
- `np.where(condição, valor_se_verdadeiro, valor_se_falso)` escolhe, posição por posição, qual valor usar. É a forma vetorizada de um `if/else` dentro de um laço.
- A condição foi aplicada sobre o **preço já reajustado**, como pede o enunciado. Por isso só os produtos 4 (R\$ 35,53) e 6 (R\$ 49,14) receberam desconto; os demais continuaram abaixo de R\$ 30,00 mesmo após o reajuste.
- Para os produtos com desconto, o aumento final foi de 1,08 × 0,95 = 1,026, ou seja, **2,6%** sobre o preço original, e não 3% (8% − 5%). Isso acontece porque os percentuais são aplicados em sequência, e o desconto incide sobre um valor já aumentado.
- `np.round(..., 2)` foi usado apenas para exibir os valores em centavos; os cálculos mantêm a precisão completa.

## Questão 11 — Identificando valores atípicos

Considere:

```python
tempos = np.array([
    42, 45, 44, 43, 46, 45, 44, 47,
    43, 44, 120, 45, 46, 42, 44
])
```

a) calcule média e mediana;  
b) compare os dois valores;  
c) identifique o valor aparentemente atípico;  
d) remova esse valor utilizando uma máscara;  
e) recalcule média e desvio padrão.

Explique em um pequeno parágrafo o efeito do valor atípico sobre a média.

````python
tempos = np.array([
    42, 45, 44, 43, 46, 45, 44, 47,
    43, 44, 120, 45, 46, 42, 44
])

# a) Média e mediana
media_tempos = np.mean(tempos)
mediana_tempos = np.median(tempos)
print(f"a) Média: {media_tempos:.2f} | Mediana: {mediana_tempos:.2f}")

# b) Comparação
print(f"b) A média é {media_tempos - mediana_tempos:.2f} unidades maior que a mediana")

# c) Valor atípico: o que mais se distancia da mediana
distancia_da_mediana = np.abs(tempos - mediana_tempos)
indice_atipico = np.argmax(distancia_da_mediana)
valor_atipico = tempos[indice_atipico]
print("c) Distância de cada valor até a mediana:", distancia_da_mediana)
print(f"   Valor atípico: {valor_atipico} (posição {indice_atipico}), a {distancia_da_mediana[indice_atipico]:.0f} unidades da mediana")

# d) Remoção com máscara
sem_atipico = tempos != valor_atipico
tempos_limpos = tempos[sem_atipico]
print("d) Tempos sem o valor atípico:", tempos_limpos)

# e) Recalcular média e desvio padrão
print(f"e) Média:         antes {media_tempos:.2f} | depois {np.mean(tempos_limpos):.2f}")
print(f"   Desvio padrão: antes {np.std(tempos):.2f} | depois {np.std(tempos_limpos):.2f}")
print(f"   Mediana:       antes {mediana_tempos:.2f} | depois {np.median(tempos_limpos):.2f}")
````

**b) Comparação:** a média (49,33) ficou 5,33 unidades acima da mediana (44,00). Como quase todos os tempos estão entre 42 e 47, a média não representa bem um valor típico do conjunto, enquanto a mediana sim.

**c) Valor atípico:** o valor **120** (posição 10). Todos os outros tempos estão no máximo a 3 unidades da mediana, enquanto 120 está a 76 unidades. - Média e mediana quase iguais (69,85 e 69,77) indicam uma distribuição **simétrica**, característica da distribuição normal.
- O intervalo de 60 a 80 corresponde a média ± 1 desvio padrão. Pela teoria, cerca de **68%** dos valores de uma distribuição normal ficam nessa faixa. Na simulação, foram 71,10%, um valor próximo do esperado.

## Questão 13 — Índices em vez dos valores

Considere:

`consumo = np.array([220, 180, 250, 195, 300, 275, 210, 185])`

Utilize `np.argmax()` e `np.argmin()` para descobrir:

a) posição do maior consumo;  
b) posição do menor consumo;  
c) seus respectivos valores.

Explique a diferença entre `np.max()` e `np.argmax()`.

````python
consumo = np.array([220, 180, 250, 195, 300, 275, 210, 185])

posicao_maior_consumo = np.argmax(consumo)
posicao_menor_consumo = np.argmin(consumo)

print(f"a) Posição do maior consumo: {posicao_maior_consumo}")
print(f"b) Posição do menor consumo: {posicao_menor_consumo}")
print(f"c) Maior consumo: {consumo[posicao_maior_consumo]} | Menor consumo: {consumo[posicao_menor_consumo]}")
print()
print("Comparação:")
print(f"   np.max(consumo)    = {np.max(consumo)}  -> o maior VALOR")
print(f"   np.argmax(consumo) = {np.argmax(consumo)}    -> a POSIÇÃO do maior valor")
````

**Diferença entre `np.max()` e `np.argmax()`:**

- `np.max(consumo)` responde **"qual é o maior valor?"** e retorna o próprio valor: `300`.
- `np.argmax(consumo)` responde **"onde está o maior valor?"** e retorna o índice dele no array: `4` (lembrando que a contagem começa em 0, então é o 5º elemento).

As duas funções se complementam: com a posição retornada por `np.argmax()` é possível obter o valor, pois `consumo[np.argmax(consumo)]` é igual a `np.max(consumo)`. O contrário não é direto, já que saber apenas o valor não informa em qual posição ele está.

Na prática, `argmax` é útil quando a posição tem significado. Por exemplo, se cada posição representasse um mês ou uma residência, `np.argmax()` diria **qual** mês ou residência teve o maior consumo, e não apenas quanto foi consumido. Foi assim que as questões 6, 7 e 8 identificaram o mês, a variação e o aluno correspondentes.

Se o valor máximo aparecer mais de uma vez, `np.argmax()` retorna apenas a posição da **primeira** ocorrência.

## Questão 14 — Ordenação

Considere as velocidades registradas:

```python
velocidades = np.array([
    82, 95, 76, 110, 88, 91, 105,
    79, 100, 86, 93, 115
])
```

Determine:

a) valores em ordem crescente;  
b) valores em ordem decrescente;  
c) três maiores velocidades;  
d) três menores velocidades;  
e) posição original da maior velocidade.

````python
velocidades = np.array([
    82, 95, 76, 110, 88, 91, 105,
    79, 100, 86, 93, 115
])

ordem_crescente = np.sort(velocidades)
ordem_decrescente = ordem_crescente[::-1]
tres_maiores = ordem_decrescente[:3]
tres_menores = ordem_crescente[:3]
posicao_maior_velocidade = np.argmax(velocidades)

print("Velocidades originais:", velocidades)
print()
print("a) Ordem crescente:  ", ordem_crescente)
print("b) Ordem decrescente:", ordem_decrescente)
print("c) Três maiores velocidades:", tres_maiores)
print("d) Três menores velocidades:", tres_menores)
print(f"e) Posição original da maior velocidade: {posicao_maior_velocidade} (valor {velocidades[posicao_maior_velocidade]})")
print()
print("Índices que ordenariam o array (np.argsort):", np.argsort(velocidades))
````

**Observações:**

- `np.sort()` retorna uma **cópia ordenada** em ordem crescente e não altera o array original. Por isso `velocidades` continua na ordem em que os dados foram registrados, e ainda é possível descobrir a posição original da maior velocidade no item e.
- O NumPy não tem um parâmetro para ordem decrescente em `np.sort()`. A solução é inverter o array já ordenado com o slicing `[::-1]`, visto na Questão 3.
- Com os arrays ordenados, os extremos saem por slicing: as três primeiras posições da ordem crescente são as menores (`[:3]`), e as três primeiras da ordem decrescente são as maiores.
- A maior velocidade (115) estava na posição **11**, a última do array original. O item e precisa ser respondido sobre `velocidades`, e não sobre um array ordenado: na ordem decrescente, por exemplo, o 115 aparece na posição 0.
- `np.argsort()` retorna os **índices** que colocariam o array em ordem crescente. O último índice (11) é a posição original do maior valor, o mesmo resultado de `np.argmax()`. Essa função é útil quando é preciso ordenar um array e manter a referência de onde cada valor estava.

## Questão 15 — Percentis

Considere:

```python
salarios = np.array([
    1800, 2100, 2300, 2500, 2700,
```

Calcule:

Explique o significado do terceiro quartil nesse conjunto de dados.

````python
salarios = np.array([
    1800, 2100, 2300, 2500, 2700,
    3000, 3200, 3500, 4000, 4500,
    5000, 6500, 8000, 12000, 18000
])

q1 = np.percentile(salarios, 25)
mediana_salarios = np.percentile(salarios, 50)
q3 = np.percentile(salarios, 75)
p90 = np.percentile(salarios, 90)

print(f"a) Primeiro quartil (Q1, 25%): R$ {q1:,.2f}")
print(f"b) Mediana (50%): R$ {mediana_salarios:,.2f}  (conferência com np.median: R$ {np.median(salarios):,.2f})")
print(f"c) Terceiro quartil (Q3, 75%): R$ {q3:,.2f}")
print(f"d) Percentil 90: R$ {p90:,.2f}")
print()
print(f"Média dos salários: R$ {np.mean(salarios):,.2f}")
print(f"Salários acima de Q3: {salarios[salarios > q3]} ({np.sum(salarios > q3)} de {salarios.size})")
````

**Observações sobre o cálculo:**

- `np.percentile(array, p)` retorna o valor abaixo do qual está `p`% dos dados. Q1, mediana e Q3 são os percentis 25, 50 e 75.
- Quando o percentil cai entre duas observações, o NumPy faz uma **interpolação linear** (método padrão). Com 15 salários ordenados, as posições vão de 0 a 14:

  - P90 fica na posição 14 × 0,90 = 12,6, entre R\$ 8.000 e R\$ 12.000 → 8.000 + 0,6 × 4.000 = **R\$ 10.400**.
- A mediana calculada com `np.percentile(salarios, 50)` coincide com `np.median()` (R\$ 3.500).

**Significado do terceiro quartil:**

O terceiro quartil de R\$ 5.750 indica que **75% dos salários são menores ou iguais a R\$ 5.750** e apenas 25% ficam acima desse valor. Neste conjunto, só 4 dos 15 salários ultrapassam o Q3 (R\$ 6.500, R\$ 8.000, R\$ 12.000 e R\$ 18.000).

O Q3 também revela a **assimetria** dos dados: a distância entre a mediana e o Q3 (R\$ 2.250) é bem maior que a distância entre o Q1 e a mediana (R\$ 900). Ou seja, os salários da metade inferior estão concentrados em uma faixa estreita, enquanto os da metade superior se espalham por valores muito mais altos.
````python
import io
import os
import zipfile

url_inmet = "https://portal.inmet.gov.br/uploads/dadoshistoricos/2025.zip"
arquivo_no_zip = "2025/INMET_N_AM_A101_MANAUS_01-01-2025_A_31-12-2025.CSV"
caminho_csv_inmet = os.path.join("dados", "INMET_A101_MANAUS_2025.CSV")

if not os.path.exists(caminho_csv_inmet):
    print("Baixando dados históricos do INMET (arquivo de ~90 MB, pode levar alguns minutos)...")
    requisicao = urllib.request.Request(url_inmet, headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(requisicao, timeout=600) as resposta:
        conteudo_zip = resposta.read()

    os.makedirs("dados", exist_ok=True)
    with zipfile.ZipFile(io.BytesIO(conteudo_zip)) as arquivo_zip:
        with open(caminho_csv_inmet, "wb") as arquivo_csv:
            arquivo_csv.write(arquivo_zip.read(arquivo_no_zip))
    print("Arquivo salvo em:", caminho_csv_inmet)
else:
    print("Usando arquivo já baixado:", caminho_csv_inmet)

# As 8 primeiras linhas do CSV descrevem a estação
print()
with open(caminho_csv_inmet, encoding="latin-1") as arquivo:
    for _ in range(8):
        print("  ", arquivo.readline().strip())
````

````python
# Colunas usadas: 0 = Data | 1 = Hora UTC | 7 = Temperatura do ar - bulbo seco (°C)
# skip_header=9 pula as 8 linhas de metadados e a linha com os nomes das colunas
dados_inmet = np.genfromtxt(caminho_csv_inmet, delimiter=";", skip_header=9,
                            usecols=(0, 1, 7), dtype=str, encoding="latin-1")

datas_horarias = dados_inmet[:, 0]

# O CSV usa vírgula como separador decimal e deixa o campo vazio quando não há medição
temperatura_texto = np.char.replace(dados_inmet[:, 2], ",", ".")
temperatura_horaria = np.where(temperatura_texto == "", "nan", temperatura_texto).astype(float)

ausentes_inmet = np.isnan(temperatura_horaria)

print("Primeiras linhas lidas:")
print(dados_inmet[:3])
print()
print("Registros horários:", temperatura_horaria.size)
print(f"Valores ausentes: {np.sum(ausentes_inmet)} ({np.mean(ausentes_inmet) * 100:.2f}% dos registros)")
````

````python
# 8.760 registros = 365 dias x 24 horas -> matriz com um dia por linha
temperatura_por_dia = temperatura_horaria.reshape(-1, 24)
datas_diarias = datas_horarias.reshape(-1, 24)[:, 0]

# Média de cada dia ignorando as horas sem medição
temperatura_media_diaria = np.nanmean(temperatura_por_dia, axis=1)

print("Matriz (dias, horas):", temperatura_por_dia.shape)
print("Observações diárias:", temperatura_media_diaria.size)
print("Dias sem nenhuma medição:", np.sum(np.isnan(temperatura_media_diaria)))
print()

media_temperatura = np.mean(temperatura_media_diaria)
indice_dia_mais_frio = np.argmin(temperatura_media_diaria)
indice_dia_mais_quente = np.argmax(temperatura_media_diaria)
amplitude_termica = np.max(temperatura_media_diaria) - np.min(temperatura_media_diaria)
dias_acima_da_media = temperatura_media_diaria > media_temperatura
desvio_temperatura = np.std(temperatura_media_diaria)

print(f"a) Temperatura média: {media_temperatura:.2f} °C")
print(f"b) Menor temperatura média diária: {temperatura_media_diaria[indice_dia_mais_frio]:.2f} °C ({datas_diarias[indice_dia_mais_frio]})")
print(f"c) Maior temperatura média diária: {temperatura_media_diaria[indice_dia_mais_quente]:.2f} °C ({datas_diarias[indice_dia_mais_quente]})")
print(f"d) Amplitude térmica: {amplitude_termica:.2f} °C")
print(f"e) Dias acima da média: {np.sum(dias_acima_da_media)} de {temperatura_media_diaria.size}")
print(f"f) Desvio padrão: {desvio_temperatura:.2f} °C")
print()
print(f"Para comparação, nos registros horários: mínima {np.nanmin(temperatura_horaria):.1f} °C | máxima {np.nanmax(temperatura_horaria):.1f} °C")
````

**Observações:**

- **Por que médias diárias:** os registros do INMET são horários, e a temperatura varia muito ao longo de um mesmo dia (noite × tarde). Para analisar o comportamento ao longo do ano, os 8.760 registros foram organizados em uma matriz de 365 linhas (dias) × 24 colunas (horas) com `reshape(-1, 24)`, e foi calculada a média de cada linha. O resultado são **365 observações**, bem acima das 30 exigidas.
- **Valores ausentes:** 73 registros horários (0,83%) estavam vazios no CSV. Eles foram convertidos para `NaN` e ignorados com `np.nanmean()`, que calcula a média apenas das horas com medição. Nenhum dia ficou totalmente sem dados, então todas as 365 médias diárias são válidas.
- **Temperatura média:** 27,60 °C, com desvio padrão de apenas 1,44 °C. Isso mostra que a temperatura média diária em Manaus varia pouco ao longo do ano, característica do clima equatorial.
- **Extremos:** o dia mais frio foi 05/07/2025 (23,27 °C de média) e o mais quente foi 03/09/2025 (31,30 °C), uma amplitude de 8,03 °C entre médias diárias. Nos registros horários a amplitude é maior (de 21,6 °C a 35,7 °C), pois inclui as variações dentro de cada dia.
- **Acima da média:** 170 dos 365 dias ficaram acima da média, um pouco menos da metade.
- As datas e horas do INMET estão em **UTC**. Manaus fica em UTC−4, então cada "dia" desta análise vai das 20h do dia anterior às 20h no horário local. Para médias diárias ao longo de um ano, esse deslocamento tem pouco efeito.

## Questão 20 — Normalização

Considere:

`dados = np.array([10, 20, 30, 40, 50, 60, 70])`

Faça a normalização Min-Max:

$$x_{\text{normalizado}} = \frac{x - \text{mínimo}}{\text{máximo} - \text{mínimo}}$$

Depois:

a) apresente o novo array;  
b) mostre seu mínimo e máximo;  
c) explique por que os valores passaram a pertencer ao intervalo [0,1].

````python
# Nome diferente de "dados" para não sobrescrever o array da Questão 3
dados_normalizacao = np.array([10, 20, 30, 40, 50, 60, 70])

minimo = np.min(dados_normalizacao)
maximo = np.max(dados_normalizacao)

# Operação vetorizada aplicada a todos os elementos
dados_normalizados = (dados_normalizacao - minimo) / (maximo - minimo)

print("Array original:  ", dados_normalizacao)
print(f"Mínimo original: {minimo} | Máximo original: {maximo}")
print()
print("a) Array normalizado:", np.round(dados_normalizados, 4))
print(f"b) Mínimo: {np.min(dados_normalizados)} | Máximo: {np.max(dados_normalizados)}")
````

**c) Por que os valores passaram a pertencer ao intervalo [0, 1]:**

A fórmula faz duas operações em sequência:

1. **Subtrai o mínimo** (`x − 10`): todos os valores são deslocados para que o menor deles vire **0**. O array passa a ser `[0, 10, 20, 30, 40, 50, 60]`, e nenhum valor fica negativo, já que nenhum elemento é menor que o mínimo.
2. **Divide pela amplitude** (`máximo − mínimo = 60`): o maior valor deslocado (60) é exatamente igual à amplitude, então vira 60 / 60 = **1**. Todos os outros valores deslocados são menores que 60, então resultam em números menores que 1.

Assim, o mínimo sempre vira 0, o máximo sempre vira 1, e todos os demais ficam entre eles, qualquer que seja a escala original dos dados.

**Observações:**

- A normalização **preserva as proporções** entre os valores: como os dados originais estavam igualmente espaçados (de 10 em 10), os normalizados também ficaram igualmente espaçados (de 1/6 ≈ 0,1667 em 1/6).
- Esse tipo de transformação é útil para **comparar variáveis de escalas diferentes**, como uma nota de 0 a 1000 no ENEM e uma temperatura em °C, colocando ambas no mesmo intervalo.
- A fórmula não funciona se todos os valores forem iguais, pois máximo − mínimo = 0 geraria uma divisão por zero.
- A normalização é sensível a valores atípicos: um único valor muito alto, como o 120 da Questão 11, viraria 1 e "espremeria" todos os outros perto de 0.

**Sobre `np.corrcoef()`:**

A função retorna uma **matriz de correlação** 2 × 2. A diagonal principal vale sempre 1, pois é a correlação de cada variável com ela mesma, e os valores fora da diagonal (posições `[0, 1]` e `[1, 0]`, iguais entre si) são a correlação entre as duas variáveis. O coeficiente de correlação de Pearson (**r**) varia de −1 a 1:

- perto de **1**: quando uma variável aumenta, a outra também aumenta, de forma quase linear;
- perto de **−1**: quando uma aumenta, a outra diminui;
- perto de **0**: não há relação linear aparente.

## Questão 27 — Correlação não significa causalidade

Crie dois arrays com pelo menos 20 observações que apresentem forte correlação.

Construa o gráfico de dispersão e calcule o coeficiente de correlação.

Depois explique:

**Uma correlação elevada é suficiente para afirmar que uma variável causa a outra?**

**Cenário simulado:** durante 30 dias de verão em uma cidade litorânea, foram registradas a **quantidade de sorvetes vendidos** e a **quantidade de resgates de afogamento** nas praias. Os dados foram gerados de propósito para que as duas variáveis dependam de uma terceira, a **temperatura do dia**, e não uma da outra: dias mais quentes levam mais pessoas a tomar sorvete e também mais pessoas à praia.

````python
np.random.seed(27)
quantidade_dias = 30

# Variável "escondida": temperatura máxima de cada dia de verão em uma cidade litorânea (°C)
temperatura_dia = np.random.uniform(24, 38, quantidade_dias)

# As duas variáveis analisadas dependem da temperatura, mas NÃO uma da outra
vendas_sorvete = 40 * temperatura_dia - 700 + np.random.normal(0, 40, quantidade_dias)
resgates_afogamento = 0.8 * temperatura_dia - 15 + np.random.normal(0, 1.5, quantidade_dias)
resgates_afogamento = np.clip(np.round(resgates_afogamento), 0, None)   # contagens inteiras e não negativas

correlacao_sorvete_resgates = np.corrcoef(vendas_sorvete, resgates_afogamento)[0, 1]
correlacao_temperatura_sorvete = np.corrcoef(temperatura_dia, vendas_sorvete)[0, 1]
correlacao_temperatura_resgates = np.corrcoef(temperatura_dia, resgates_afogamento)[0, 1]

print("Observações:", quantidade_dias, "dias")
print(f"Vendas de sorvete: de {np.min(vendas_sorvete):.0f} a {np.max(vendas_sorvete):.0f} unidades por dia")
print(f"Resgates de afogamento: de {np.min(resgates_afogamento):.0f} a {np.max(resgates_afogamento):.0f} por dia")
print()
print(f"Correlação sorvete × resgates:     r = {correlacao_sorvete_resgates:.4f}")
print(f"Correlação temperatura × sorvete:  r = {correlacao_temperatura_sorvete:.4f}")
print(f"Correlação temperatura × resgates: r = {correlacao_temperatura_resgates:.4f}")
````

````python
from matplotlib.colors import LinearSegmentedColormap

# Escala sequencial de um único tom (azul claro = dias mais frescos, azul escuro = dias mais quentes)
escala_azul = LinearSegmentedColormap.from_list("azul", ["#b7d3f6", "#2a78d6", "#0d366b"])

fig, ax = plt.subplots()
pontos = ax.scatter(vendas_sorvete, resgates_afogamento, c=temperatura_dia, cmap=escala_azul,
                    s=70, edgecolor="white", linewidth=1.5, zorder=3)

barra_cores = fig.colorbar(pontos, ax=ax, pad=0.02)
barra_cores.set_label("Temperatura máxima do dia (°C)", color=COR_TEXTO_SECUNDARIO)
barra_cores.outline.set_visible(False)

ax.text(0.02, 0.95, f"r = {correlacao_sorvete_resgates:.2f}".replace(".", ","), transform=ax.transAxes,
        fontsize=12, fontweight="bold", color=COR_TEXTO_SECUNDARIO, va="top")

ax.set_title("Vendas de sorvete × resgates de afogamento (30 dias de verão)")
ax.set_xlabel("Sorvetes vendidos no dia (unidades)")
ax.set_ylabel("Resgates de afogamento no dia")
ax.grid(axis="x", color="#e1e0d9", linewidth=0.8)
plt.show()
````

**Resultado:** o coeficiente de correlação entre vendas de sorvete e resgates de afogamento é **r = 0,88**, uma correlação positiva **forte**. No gráfico, os pontos sobem da esquerda para a direita: nos dias com mais sorvetes vendidos, houve mais resgates.

**Uma correlação elevada é suficiente para afirmar que uma variável causa a outra?**

**Não.** A correlação indica apenas que duas variáveis **variam juntas**, e não por que isso acontece. Neste exemplo, seria absurdo concluir que tomar sorvete provoca afogamentos, ou que proibir a venda de sorvetes reduziria os resgates. A explicação está em uma **variável de confusão**, a temperatura:

- temperatura × vendas de sorvete: r = 0,97;
- temperatura × resgates de afogamento: r = 0,92.

A cor dos pontos no gráfico deixa isso visível: os pontos claros (dias mais frescos) ficam no canto inferior esquerdo, com poucas vendas e poucos resgates, e os escuros (dias mais quentes) no canto superior direito. O calor aumenta as duas variáveis ao mesmo tempo, e isso cria a correlação entre elas, mesmo sem nenhuma relação de causa direta.

Em geral, uma correlação alta entre A e B pode ter várias explicações:

1. **A causa B**;
2. **B causa A** (causalidade reversa);
3. **uma terceira variável causa A e B** (como a temperatura aqui);
4. **coincidência**, principalmente em amostras pequenas ou quando se testam muitas combinações de variáveis.

Para afirmar causalidade, é preciso mais do que um coeficiente: por exemplo, um experimento controlado, em que apenas uma variável é alterada de propósito, ou uma análise que leve em conta as possíveis variáveis de confusão. Isso também vale para a Questão 26: a forte correlação entre horas de estudo e notas é compatível com o estudo melhorar as notas, mas, sozinha, não prova isso.

## Questão 29 — Média móvel

Considere:

```python
vendas = np.array([
    100, 105, 98, 110, 115, 120,
    118, 125, 130, 128, 140, 145,
    150, 148, 155, 160, 170, 175
])
```

Calcule uma **média móvel de três períodos**.

Sugestão: pesquise o funcionamento de `np.convolve()`.

Construa um gráfico contendo:

- série original;
- média móvel.

Explique o efeito da média móvel sobre as oscilações da série.

````python
# Nome diferente de "vendas" para não sobrescrever o array das Questões 6 e 21
vendas_periodos = np.array([
    100, 105, 98, 110, 115, 120,
    118, 125, 130, 128, 140, 145,
    150, 148, 155, 160, 170, 175
])
periodos = np.arange(1, vendas_periodos.size + 1)

janela = 3
pesos = np.ones(janela) / janela   # [1/3, 1/3, 1/3]

# mode="valid": só calcula onde a janela de 3 valores cabe inteira -> 18 - 3 + 1 = 16 médias
media_movel = np.convolve(vendas_periodos, pesos, mode="valid")

# Cada média usa o período atual e os 2 anteriores, então a primeira corresponde ao período 3
periodos_media_movel = periodos[janela - 1:]

print("Pesos da janela:", np.round(pesos, 4))
print("Série original:", vendas_periodos, f"({vendas_periodos.size} valores)")
print("Média móvel:   ", np.round(media_movel, 2), f"({media_movel.size} valores)")
print()
print(f"Conferência do 1º valor: (100 + 105 + 98) / 3 = {(100 + 105 + 98) / 3:.2f}")
print()

variacoes_originais = np.diff(vendas_periodos)
variacoes_media_movel = np.diff(media_movel)
print(f"Quedas na série original: {np.sum(variacoes_originais < 0)} | quedas na média móvel: {np.sum(variacoes_media_movel < 0)}")
print(f"Desvio padrão das variações: original {np.std(variacoes_originais):.2f} | média móvel {np.std(variacoes_media_movel):.2f}")
````

**Como `np.convolve()` calcula a média móvel:**

A convolução "desliza" um pequeno array de pesos sobre a série e, em cada posição, soma os valores multiplicados pelos pesos. Usando os pesos `[1/3, 1/3, 1/3]`, cada resultado é exatamente a média de 3 valores consecutivos:

- 1ª média: (100 + 105 + 98) / 3 = 101,00
- 2ª média: (105 + 98 + 110) / 3 = 104,33
- e assim por diante, avançando um período por vez.

Com `mode="valid"`, o cálculo é feito apenas onde a janela cabe inteira na série. Por isso, 18 valores geram **16 médias**: não existe média de 3 períodos para os períodos 1 e 2. Cada média foi posicionada no **último** período da sua janela (a média dos períodos 1, 2 e 3 aparece no período 3), de modo que ela usa apenas valores já observados até aquele ponto.

````python
fig, ax = plt.subplots()
ax.plot(periodos, vendas_periodos, marker="o", color=CORES[0], label="Série original")
ax.plot(periodos_media_movel, media_movel, color=CORES[1], linewidth=2.5,
        label="Média móvel de 3 períodos")

ax.set_title("Vendas: série original × média móvel de 3 períodos")
ax.set_xlabel("Período")
ax.set_ylabel("Vendas")
ax.set_xticks(periodos)
ax.legend(loc="upper left")
plt.show()
````

**Efeito da média móvel sobre as oscilações da série:**

A média móvel **suaviza** a série. Como cada ponto é a média de 3 períodos, um valor que sobe ou desce fora do padrão tem seu efeito dividido com os vizinhos, e as oscilações de curto prazo são amortecidas. No gráfico, a linha azul (original) tem vários "dentes", enquanto a linha laranja (média móvel) sobe de forma quase contínua.

Os números confirmam:

- a série original tem **4 quedas** (períodos 3, 7, 10 e 14), e a média móvel **não tem nenhuma**: nela, as vendas crescem em todos os períodos;
- o desvio padrão das variações entre períodos consecutivos cai de **4,95** na série original para **1,74** na média móvel, ou seja, as mudanças de um período para o outro ficam bem mais regulares.

Com isso, a **tendência de crescimento** fica mais fácil de enxergar, sem ser confundida com quedas pontuais, como a de 105 para 98 no período 3.

A suavização também tem custos:

- **atraso:** como a média usa valores passados, ela reage com atraso às mudanças. Em uma série crescente, a média móvel fica quase sempre **abaixo** da série original, como se vê no gráfico, principalmente no final, quando as vendas aceleram (160 → 170 → 175);
- **perda de pontos:** os dois primeiros períodos ficam sem média;
- **escolha da janela:** janelas maiores suavizam mais, mas aumentam o atraso e escondem mais detalhes.

## Questão 31 — Detecção de outliers com IQR

Gere os dados:

```python
np.random.seed(30)

dados = np.random.normal(100, 10, 200)
dados = np.append(dados, [40, 45, 160, 170])
```

Utilizando o método do intervalo interquartil:

- IQR = Q3 − Q1
- Limite inferior = Q1 − 1,5 × IQR
- Limite superior = Q3 + 1,5 × IQR

identifique os possíveis outliers.

Construa um boxplot antes e depois da remoção dos outliers.

Compare média e desvio padrão.

````python
np.random.seed(30)
# Nome diferente de "dados" para não sobrescrever o array da Questão 3
dados_outliers = np.random.normal(100, 10, 200)
dados_outliers = np.append(dados_outliers, [40, 45, 160, 170])

q1_iqr, q3_iqr = np.percentile(dados_outliers, [25, 75])
iqr = q3_iqr - q1_iqr
limite_inferior = q1_iqr - 1.5 * iqr
limite_superior = q3_iqr + 1.5 * iqr

eh_outlier = (dados_outliers < limite_inferior) | (dados_outliers > limite_superior)
outliers = dados_outliers[eh_outlier]
dados_sem_outliers = dados_outliers[~eh_outlier]

print(f"Total de observações: {dados_outliers.size}")
print(f"Q1 = {q1_iqr:.2f} | Q3 = {q3_iqr:.2f} | IQR = {iqr:.2f}")
print(f"Limite inferior = Q1 - 1,5 × IQR = {limite_inferior:.2f}")
print(f"Limite superior = Q3 + 1,5 × IQR = {limite_superior:.2f}")
print()
print(f"Possíveis outliers ({outliers.size}):", np.round(np.sort(outliers), 2))
print(f"Observações após a remoção: {dados_sem_outliers.size}")
print()
print(f"Média:         antes {np.mean(dados_outliers):.2f} | depois {np.mean(dados_sem_outliers):.2f}")
print(f"Desvio padrão: antes {np.std(dados_outliers):.2f} | depois {np.std(dados_sem_outliers):.2f}")
print(f"Mediana:       antes {np.median(dados_outliers):.2f} | depois {np.median(dados_sem_outliers):.2f}")
````

````python
fig, (ax_antes, ax_depois) = plt.subplots(1, 2, figsize=(12, 5), sharey=True)

estilo_boxplot = dict(
    widths=0.5,
    patch_artist=True,
    boxprops={"facecolor": CORES[0], "edgecolor": CORES[0]},
    medianprops={"color": "white", "linewidth": 2},
    whiskerprops={"color": COR_TEXTO_SECUNDARIO},
    capprops={"color": COR_TEXTO_SECUNDARIO},
    flierprops={"marker": "o", "markersize": 7, "markerfacecolor": "none",
                "markeredgecolor": CORES[1], "markeredgewidth": 1.5},
)

ax_antes.boxplot(dados_outliers, **estilo_boxplot)
ax_antes.set_title(f"Antes da remoção ({dados_outliers.size} valores)", fontsize=12)

ax_depois.boxplot(dados_sem_outliers, **estilo_boxplot)
ax_depois.set_title(f"Depois da remoção ({dados_sem_outliers.size} valores)", fontsize=12)

for ax in (ax_antes, ax_depois):
    ax.axhline(limite_inferior, color=COR_REFERENCIA, linewidth=1, linestyle="--")
    ax.axhline(limite_superior, color=COR_REFERENCIA, linewidth=1, linestyle="--")
    ax.set_xticks([1], ["Dados"])
    ax.set_xlabel("Conjunto de dados")

ax_antes.set_ylabel("Valor")
ax_antes.text(1.3, limite_superior + 2, f"Limite superior ({limite_superior:.1f})".replace(".", ","),
              fontsize=9, color=COR_TEXTO_SECUNDARIO)
ax_antes.text(1.3, limite_inferior - 6, f"Limite inferior ({limite_inferior:.1f})".replace(".", ","),
              fontsize=9, color=COR_TEXTO_SECUNDARIO)

fig.suptitle("Detecção de outliers pelo método do IQR", x=0.01, ha="left", fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
````

**Identificação dos outliers:**

- Q1 = 92,33 e Q3 = 107,47, logo **IQR = 15,14**. Os 50% centrais dos dados ocupam uma faixa de pouco mais de 15 unidades.
- Limite inferior = 92,33 − 1,5 × 15,14 = **69,62**
- Limite superior = 107,47 + 1,5 × 15,14 = **130,18**

Qualquer valor fora da faixa [69,62; 130,18] é considerado um possível outlier. A máscara `(dados < limite_inferior) | (dados > limite_superior)` encontrou **5 valores**: **40, 45, 160 e 170**, os 4 valores extremos acrescentados de propósito, e **130,2**, um valor gerado pela própria distribuição normal que ficou apenas 0,02 acima do limite superior.

**Boxplots:** no gráfico "antes", os 5 outliers aparecem como círculos laranja além dos bigodes e das linhas tracejadas dos limites. No gráfico "depois", eles não existem mais, e todos os dados ficam dentro dos limites. A caixa (Q1 a Q3) e a mediana praticamente não mudam, pois dependem da posição dos valores centrais e não dos extremos.

**Comparação de média e desvio padrão:**

| | Observações | Média | Desvio padrão |
|---|---|---|---|
| Antes | 204 | 99,98 | 13,39 |
| Depois | 199 | 99,75 | 10,15 |

- **A média quase não mudou** (−0,23). Isso acontece porque os outliers estão dos **dois lados**: os valores baixos (40 e 45) puxavam a média para baixo e os altos (130,2, 160 e 170) para cima, e esses efeitos praticamente se cancelavam. **Cuidados:**

- O método do IQR aponta **possíveis** outliers, não erros certos. O valor 130,2 foi gerado pela mesma distribuição dos demais, é apenas um valor alto e legítimo. Antes de remover um outlier em dados reais, é preciso investigar se ele é um erro de medição ou digitação ou uma observação real e importante.
- Depois da remoção, os quartis e os limites mudam um pouco (o novo limite superior seria 129,42). Mesmo assim, nenhum dos 199 valores restantes fica fora dos novos limites.

## Questão 32 — Crescimento acumulado

Considere os retornos mensais percentuais de um investimento:

```python

    2.0, -1.0, 3.0, 1.5,
```

````python
retornos = np.array([
    2.0, -1.0, 3.0, 1.5,
    -2.0, 4.0, 2.5, -0.5,
    3.0, 1.0, -1.5, 2.0
])
investimento_inicial = 1000.00

# Retorno de 2% -> fator 1,02 | retorno de -1% -> fator 0,99
fatores = 1 + retornos / 100

# np.cumprod multiplica os fatores acumulando: f1, f1*f2, f1*f2*f3, ...
fator_acumulado = np.cumprod(fatores)
valor_investimento = investimento_inicial * fator_acumulado

# Série com o mês 0 (valor inicial) para o gráfico
meses_investimento = np.arange(0, retornos.size + 1)
valores_com_inicio = np.concatenate(([investimento_inicial], valor_investimento))

print("Fatores mensais:   ", fatores)
print("Fator acumulado:   ", np.round(fator_acumulado, 4))
print()
print("Mês | Retorno | Valor do investimento")
for mes, retorno, valor in zip(meses_investimento[1:], retornos, valor_investimento):   # laço apenas para exibir a tabela
    print(f"{mes:>3} | {retorno:>+6.1f}% | R$ {valor:>9,.2f}")
print()

rentabilidade_total = (fator_acumulado[-1] - 1) * 100
soma_simples = np.sum(retornos)
print(f"Valor final: R$ {valor_investimento[-1]:,.2f}")
print(f"Rentabilidade acumulada: {rentabilidade_total:.2f}%")
print(f"Soma simples dos retornos (para comparação): {soma_simples:.2f}%")
print(f"Maior valor: R$ {np.max(valor_investimento):,.2f} no mês {np.argmax(valor_investimento) + 1}")
````

````python
fig, ax = plt.subplots()
ax.plot(meses_investimento, valores_com_inicio, marker="o", color=CORES[0])
ax.axhline(investimento_inicial, color=COR_REFERENCIA, linewidth=1.5, linestyle="--")
ax.text(12, investimento_inicial - 8, "Investimento inicial: R$ 1.000,00",
        ha="right", va="top", fontsize=10, color=COR_TEXTO_SECUNDARIO)

# Destaque dos meses com retorno negativo
meses_negativos = meses_investimento[1:][retornos < 0]
ax.scatter(meses_negativos, valor_investimento[retornos < 0], s=90, color="white",
           edgecolor=CORES[7], linewidth=2.5, zorder=3, label="Mês com retorno negativo")

ax.annotate(f"R$ {valor_investimento[-1]:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."),
            xy=(12, valor_investimento[-1]), xytext=(4, 12), textcoords="offset points",
            ha="right", fontsize=10, fontweight="bold", color=COR_TEXTO_SECUNDARIO)

ax.set_title("Evolução de um investimento de R$ 1.000,00")
ax.set_xlabel("Mês")
ax.set_ylabel("Valor do investimento (R$)")
ax.set_xticks(meses_investimento)
ax.set_ylim(960, 1200)
ax.yaxis.set_major_formatter(lambda valor, _: f"{valor:,.0f}".replace(",", "."))
ax.legend(loc="upper left")
plt.show()
````

**Como o cálculo foi feito:**

1. Cada retorno percentual foi convertido em um **fator de multiplicação**: `1 + retorno / 100`. Um retorno de +2% vira 1,02 (o valor é multiplicado por 1,02), e um retorno de −1% vira 0,99.
2. `np.cumprod()` calcula o **produto acumulado** desses fatores: o 1º elemento é f₁, o 2º é f₁ × f₂, o 3º é f₁ × f₂ × f₃, e assim por diante. Cada posição indica quanto o valor inicial foi multiplicado até aquele mês.
3. Multiplicando o fator acumulado por R\$ 1.000,00, obtém-se o valor do investimento em cada mês, tudo de forma vetorizada.

Por exemplo, no mês 2: 1.000 × 1,02 × 0,99 = **R\$ 1.009,80**.

**Resultados:**

- **Valor final** após 12 meses: **R\$ 1.146,94**
- **Rentabilidade acumulada:** **14,69%**
- O investimento terminou no seu maior valor (mês 12) e ficou acima do valor inicial em todos os meses.

**Observações:**

- **Por que multiplicar e não somar:** somando os retornos, o resultado seria 14,00%, mas o ganho real foi 14,69%. Isso acontece por causa dos **juros compostos**: o retorno de cada mês incide sobre o valor já acumulado, e não sobre os R\$ 1.000,00 iniciais. Os ganhos de um mês também passam a render nos meses seguintes.
- **Perdas e recuperação:** os 4 meses com retorno negativo (2, 5, 8 e 11, destacados em vermelho no gráfico) reduziram o valor, mas os retornos positivos seguintes recuperaram as perdas. A maior queda foi no mês 5 (−2%, de R\$ 1.055,70 para R\$ 1.034,58), recuperada logo no mês 6 com o maior retorno do período (+4%).
- O efeito composto também vale para as perdas: uma queda de 2% seguida de uma alta de 2% **não** devolve o valor original, porque 0,98 × 1,02 = 0,9996.

````python
# Reaproveita o CSV da estação A101 (Manaus, 2025) lido na Questão 18
temperatura_bruta = temperatura_horaria.copy()
horas_utc = np.char.replace(dados_inmet[:, 1], " UTC", "").astype(int) // 100
hora_local = (horas_utc - 4) % 24                      # Manaus fica em UTC-4
mes_registro = np.char.partition(datas_horarias, "/")[:, 2]
mes_registro = np.char.partition(mes_registro, "/")[:, 0].astype(int)

ausentes_35 = np.isnan(temperatura_bruta)
posicoes_horas = np.arange(temperatura_bruta.size)

print(f"Registros horários: {temperatura_bruta.size}")
print(f"a) Valores ausentes: {np.sum(ausentes_35)} ({np.mean(ausentes_35) * 100:.2f}%)")

# Tamanho dos blocos de horas seguidas sem medição
posicoes_ausentes = posicoes_horas[ausentes_35]
blocos = np.split(posicoes_ausentes, np.where(np.diff(posicoes_ausentes) != 1)[0] + 1)
tamanhos_blocos = np.array([bloco.size for bloco in blocos])
print(f"   Blocos de ausência: {tamanhos_blocos.size} | maior bloco: {np.max(tamanhos_blocos)} horas seguidas")

# Tratamento: interpolação linear entre a medição anterior e a seguinte
temperatura_tratada = temperatura_bruta.copy()
temperatura_tratada[ausentes_35] = np.interp(posicoes_horas[ausentes_35],
                                             posicoes_horas[~ausentes_35],
                                             temperatura_bruta[~ausentes_35])

print(f"   Ausentes após o tratamento: {np.sum(np.isnan(temperatura_tratada))}")
print(f"   Observações válidas: {temperatura_tratada.size}")
print()
# Exemplo: o maior bloco de ausência, com uma medição antes e uma depois
inicio_maior_bloco = blocos[np.argmax(tamanhos_blocos)][0]
trecho = slice(inicio_maior_bloco - 1, inicio_maior_bloco + np.max(tamanhos_blocos) + 1)
print(f"Maior bloco tratado ({datas_horarias[inicio_maior_bloco]}, a partir de {dados_inmet[inicio_maior_bloco, 1]}):")
print("   antes: ", temperatura_bruta[trecho])
print("   depois:", np.round(temperatura_tratada[trecho], 2))
````

**a) Tratamento dos valores ausentes:**

Dos 8.760 registros horários, **73 (0,83%) estavam vazios**. Eles estão espalhados em **61 blocos curtos**: a maioria é de 1 hora isolada, e o maior tem 6 horas seguidas (4 de novembro).

Como a temperatura muda de forma gradual de uma hora para a outra, os ausentes foram preenchidos por **interpolação linear** com `np.interp()`. Cada valor faltante é estimado sobre uma reta que liga a última medição antes da falha à primeira medição depois dela. No maior bloco, por exemplo, a temperatura caiu de 32,0 °C para 26,1 °C, e as 6 horas ausentes foram preenchidas com valores decrescentes entre esses dois pontos.

Esse método foi preferido à substituição pela média usada na Questão 30. Em uma série horária, trocar uma hora ausente da tarde pela média anual (27,6 °C) criaria "quedas" artificiais, enquanto a interpolação respeita a tendência local. Depois do tratamento, os **8.760 registros são válidos**, e média, mediana e desvio padrão praticamente não mudaram em relação ao cálculo que apenas ignora os `NaN`.

---

**Dados utilizados:** a amostra de 20.000 participantes do ENEM 2024 gerada na Questão 19 (`dados/amostra_enem_2024.csv`). O enunciado pede uma área; como a amostra contém todas, a análise foi feita para as **cinco áreas**, o que permite comparar as distribuições entre si: Ciências da Natureza (CN), Ciências Humanas (CH), Linguagens e Códigos (LC), Matemática (MT) e Redação.

**Critérios adotados:**

- **Registros sem nota:** campo vazio (participante faltou ou foi eliminado) foi removido.
- **Nota zero:** também foi removida. Nas provas objetivas, é um valor fora da escala da TRI, como discutido na Questão 19. Na Redação, todas as 746 notas zero correspondem a redações com status diferente de "sem problemas" (em branco, fuga ao tema, anulada etc.), ou seja, não são uma avaliação do texto.

````python
# 1. Seleção das notas de cada área | 2. Eliminação dos registros sem nota
colunas_areas = ["NU_NOTA_CN", "NU_NOTA_CH", "NU_NOTA_LC", "NU_NOTA_MT", "NU_NOTA_REDACAO"]
nomes_areas = ["Ciências da Natureza", "Ciências Humanas", "Linguagens", "Matemática", "Redação"]
siglas_areas = ["CN", "CH", "LC", "MT", "Redação"]

def notas_validas(coluna):
    """Converte a coluna em float e remove notas ausentes (campo vazio) e iguais a zero."""
    texto = dados_enem[:, indice_coluna[coluna]]
    notas = np.where(texto == "", "nan", texto).astype(float)
    return notas[~np.isnan(notas) & (notas > 0)]

notas_por_area = []
for coluna, nome in zip(colunas_areas, nomes_areas):   # uma iteração por área (5 colunas)
    texto = dados_enem[:, indice_coluna[coluna]]
    sem_nota = np.sum(texto == "")
    notas_zero = np.sum(texto == "0") + np.sum(texto == "0.0")
    validas = notas_validas(coluna)
    notas_por_area.append(validas)
    print(f"{nome:<21} sem nota: {sem_nota:>5} | nota zero: {notas_zero:>4} | notas válidas: {validas.size:>6}")
````

````python
# 3. Média, mediana, quartis e desvio padrão | 4. Percentual por faixa de nota
faixas = np.array([0, 400, 600, 800, 1000.01])   # [0, 400) [400, 600) [600, 800) [800, 1000]

print(f"{'Área':<21} {'Média':>7} {'Mediana':>8} {'Q1':>7} {'Q3':>7} {'Desvio':>7} | {'<400':>6} {'400-600':>8} {'600-800':>8} {'>=800':>6}")
for nome, notas in zip(nomes_areas, notas_por_area):   # laço apenas para montar a tabela
    q1_area, q3_area = np.percentile(notas, [25, 75])
    contagem_faixas, _ = np.histogram(notas, bins=faixas)
    percentuais = contagem_faixas / notas.size * 100
    print(f"{nome:<21} {np.mean(notas):>7.2f} {np.median(notas):>8.2f} {q1_area:>7.2f} {q3_area:>7.2f} {np.std(notas):>7.2f} | "
          f"{percentuais[0]:>5.2f}% {percentuais[1]:>7.2f}% {percentuais[2]:>7.2f}% {percentuais[3]:>5.2f}%")
````

````python
# Rede da escola do ensino médio: 1 Federal, 2 Estadual, 3 Municipal, 4 Privada (vazio = não informado)
rede_escola = dados_enem[:, indice_coluna["TP_DEPENDENCIA_ADM_ESC"]]

escola_publica = np.isin(rede_escola, ["1", "2", "3"])
escola_privada = rede_escola == "4"

codigos_rede, quantidades_rede = np.unique(rede_escola, return_counts=True)
print("Códigos de rede:", codigos_rede, "(vazio = não informado)")
print("Quantidades:    ", quantidades_rede)
print()
print(f"Escola pública: {np.sum(escola_publica)} participantes | Escola privada: {np.sum(escola_privada)} participantes")
print(f"Sem informação de escola (fora da comparação): {np.sum(~escola_publica & ~escola_privada)}")

def notas_do_grupo(coluna, mascara_grupo):
    """Notas válidas (não vazias e maiores que zero) dos participantes de um grupo."""
    texto = dados_enem[:, indice_coluna[coluna]]
    notas = np.where(texto == "", "nan", texto).astype(float)
    return notas[mascara_grupo & ~np.isnan(notas) & (notas > 0)]

notas_publica = [notas_do_grupo(coluna, escola_publica) for coluna in colunas_areas]
notas_privada = [notas_do_grupo(coluna, escola_privada) for coluna in colunas_areas]
````

````python
def resumo(notas):
    q1_grupo, q3_grupo = np.percentile(notas, [25, 75])
    return notas.size, np.mean(notas), np.median(notas), np.std(notas), q1_grupo, q3_grupo

print(f"{'Área':<21} {'Grupo':<8} {'n':>5} {'Média':>7} {'Mediana':>8} {'Desvio':>7} {'Q1':>7} {'Q3':>7}")
for nome, publica, privada in zip(nomes_areas, notas_publica, notas_privada):   # laço apenas para montar a tabela
    for grupo, notas in [("Pública", publica), ("Privada", privada)]:
        n, media, mediana, desvio, q1_grupo, q3_grupo = resumo(notas)
        print(f"{nome:<21} {grupo:<8} {n:>5} {media:>7.2f} {mediana:>8.2f} {desvio:>7.2f} {q1_grupo:>7.2f} {q3_grupo:>7.2f}")
    diferenca_medianas = np.median(privada) - np.median(publica)
    privada_acima = np.mean(privada > np.median(publica)) * 100
    print(f"{'':<21} Diferença das medianas (privada − pública): {diferenca_medianas:+.2f} | "
          f"privada acima da mediana da pública: {privada_acima:.1f}%")
    print()
````

**Fonte dos dados:**

- **Arquivo original:** `microdados_ed_basica_2024.csv`, uma linha por escola do Brasil (215.545 escolas, 426 colunas)
- **Acesso:** <https://www.gov.br/inep/pt-br/acesso-a-informacao/dados-abertos/microdados/censo-escolar> (arquivo `microdados_censo_escolar_2024.zip`)
- **Extrato utilizado:** as **6.181 escolas do Amazonas**, com 14 colunas, geradas pelo script [`dados/gerar_extrato_censo_escolar.py`](dados/gerar_extrato_censo_escolar.py) e salvas em `dados/censo_escolar_2024_am.csv`

**Variável quantitativa principal:** `QT_MAT_BAS`, a quantidade de **matrículas na educação básica** de cada escola. Também foram usadas a localização da escola (urbana ou rural), o município e indicadores de infraestrutura (internet, biblioteca e energia elétrica).

2. **Quanto das matrículas da educação básica do estado está concentrado em Manaus?**
3. **A infraestrutura das escolas rurais é diferente da das escolas urbanas?**

````python
caminho_censo = os.path.join("dados", "censo_escolar_2024_am.csv")

colunas_censo = np.genfromtxt(caminho_censo, delimiter=";", max_rows=1, dtype=str, encoding="utf-8")
dados_censo = np.genfromtxt(caminho_censo, delimiter=";", skip_header=1, dtype=str, encoding="utf-8")
indice_censo = {nome: posicao for posicao, nome in enumerate(colunas_censo)}

print("Escolas do Amazonas no Censo Escolar 2024:", dados_censo.shape[0])
print("Colunas:", colunas_censo)
print()

# Filtragem 1: apenas escolas em atividade (1 = em atividade, 2 = paralisada, 3 = extinta)
situacao = dados_censo[:, indice_censo["TP_SITUACAO_FUNCIONAMENTO"]]
codigos_situacao, quantidades_situacao = np.unique(situacao, return_counts=True)
print("Situação de funcionamento:", codigos_situacao, quantidades_situacao)
escolas = dados_censo[situacao == "1"]

def coluna_numerica(nome):
    """Converte uma coluna do extrato em float, com NaN nos campos vazios."""
    texto = escolas[:, indice_censo[nome]]
    return np.where(texto == "", "nan", texto).astype(float)

municipio_escola = escolas[:, indice_censo["NO_MUNICIPIO"]]
localizacao = escolas[:, indice_censo["TP_LOCALIZACAO"]]
matriculas = coluna_numerica("QT_MAT_BAS")
tem_internet = coluna_numerica("IN_INTERNET")
tem_biblioteca = coluna_numerica("IN_BIBLIOTECA")
sem_energia = coluna_numerica("IN_ENERGIA_INEXISTENTE")

# Filtragem 2: escolas com número de matrículas informado
matricula_informada = ~np.isnan(matriculas)

urbana = localizacao == "1"
rural = localizacao == "2"

print(f"Escolas em atividade: {escolas.shape[0]} (urbanas: {np.sum(urbana)} | rurais: {np.sum(rural)})")
print(f"Escolas em atividade sem matrículas informadas: {np.sum(~matricula_informada)}")
````

### Pergunta 1 — As escolas rurais atendem menos alunos que as urbanas?

Estatísticas de matrículas por escola em cada grupo (filtragem por localização) e boxplot comparando os dois grupos.

````python
matriculas_urbanas = matriculas[urbana & matricula_informada]
matriculas_rurais = matriculas[rural & matricula_informada]

print(f"{'Grupo':<8} {'Escolas':>7} {'Média':>8} {'Mediana':>8} {'Desvio':>8} {'Q1':>6} {'Q3':>6} {'IQR':>6} {'Mín':>5} {'Máx':>6}")
for grupo, valores in [("Urbana", matriculas_urbanas), ("Rural", matriculas_rurais)]:   # laço apenas para a tabela
    q1_grupo, q3_grupo = np.percentile(valores, [25, 75])
    print(f"{grupo:<8} {valores.size:>7} {np.mean(valores):>8.1f} {np.median(valores):>8.1f} {np.std(valores):>8.1f} "
          f"{q1_grupo:>6.0f} {q3_grupo:>6.0f} {q3_grupo - q1_grupo:>6.0f} {np.min(valores):>5.0f} {np.max(valores):>6.0f}")
print()
print(f"Escolas com menos de 50 alunos: urbanas {np.mean(matriculas_urbanas < 50) * 100:.1f}% | rurais {np.mean(matriculas_rurais < 50) * 100:.1f}%")
print(f"Participação das escolas rurais: {np.mean(rural) * 100:.1f}% das escolas e "
      f"{np.sum(matriculas_rurais) / np.nansum(matriculas) * 100:.1f}% das matrículas")

fig, ax = plt.subplots()
ax.boxplot([matriculas_urbanas, matriculas_rurais], tick_labels=["Urbana", "Rural"], widths=0.5, patch_artist=True,
           boxprops={"facecolor": CORES[0], "edgecolor": CORES[0]},
           medianprops={"color": "white", "linewidth": 2},
           whiskerprops={"color": COR_TEXTO_SECUNDARIO}, capprops={"color": COR_TEXTO_SECUNDARIO},
           flierprops={"marker": "o", "markersize": 3, "markerfacecolor": "none",
                       "markeredgecolor": COR_TEXTO_SECUNDARIO, "alpha": 0.5})
for posicao, valores in enumerate([matriculas_urbanas, matriculas_rurais], start=1):
    ax.text(posicao + 0.3, np.median(valores), f"mediana: {np.median(valores):.0f}".replace(".", ","),
            va="center", fontsize=10, color=COR_TEXTO_SECUNDARIO)
ax.set_xlim(0.5, 2.9)
ax.set_title("P1. Matrículas por escola: urbanas × rurais (Amazonas, 2024)")
ax.set_xlabel("Localização da escola")
ax.set_ylabel("Matrículas na educação básica")
plt.show()
````

### Pergunta 2 — Quanto das matrículas está concentrado em Manaus?

A **agregação** por município foi feita de forma vetorizada: `np.unique(..., return_inverse=True)` associa cada escola ao índice do seu município, e `np.bincount(indices, weights=matriculas)` soma as matrículas de todas as escolas de cada município de uma só vez.

````python
# Agregação: soma de matrículas e contagem de escolas por município
nomes_municipios_censo, indice_municipio = np.unique(municipio_escola[matricula_informada], return_inverse=True)
matriculas_por_municipio = np.bincount(indice_municipio, weights=matriculas[matricula_informada])
escolas_por_municipio = np.bincount(indice_municipio)

total_matriculas = np.sum(matriculas_por_municipio)
ordem_municipios = np.argsort(matriculas_por_municipio)[::-1]
eh_manaus = nomes_municipios_censo == "Manaus"

print(f"Municípios: {nomes_municipios_censo.size} | Total de matrículas no estado: {total_matriculas:,.0f}")
print(f"Manaus: {matriculas_por_municipio[eh_manaus][0]:,.0f} matrículas "
      f"({matriculas_por_municipio[eh_manaus][0] / total_matriculas * 100:.1f}% do estado) em "
      f"{escolas_por_municipio[eh_manaus][0]} escolas ({escolas_por_municipio[eh_manaus][0] / np.sum(escolas_por_municipio) * 100:.1f}% das escolas)")
print(f"2º município ({nomes_municipios_censo[ordem_municipios[1]]}): {matriculas_por_municipio[ordem_municipios[1]] / total_matriculas * 100:.1f}% das matrículas")
print(f"Mediana de matrículas por município: {np.median(matriculas_por_municipio):,.0f}")
print()
print("10 municípios com mais matrículas:")
for posicao in ordem_municipios[:10]:   # laço apenas para exibir
    print(f"  {nomes_municipios_censo[posicao]:<26} {matriculas_por_municipio[posicao]:>9,.0f}  "
          f"({matriculas_por_municipio[posicao] / total_matriculas * 100:4.1f}%)")

top10_nomes = nomes_municipios_censo[ordem_municipios[:10]]
top10_percentual = matriculas_por_municipio[ordem_municipios[:10]] / total_matriculas * 100

fig, ax = plt.subplots(figsize=(10, 5.5))
posicoes_barras = np.arange(10)[::-1]
ax.barh(posicoes_barras, top10_percentual, height=0.7, color=CORES[0])
for posicao, percentual in zip(posicoes_barras, top10_percentual):
    ax.text(percentual + 0.5, posicao, f"{percentual:.1f}%".replace(".", ","), va="center",
            fontsize=10, color=COR_TEXTO_SECUNDARIO)
ax.set_yticks(posicoes_barras, top10_nomes)
ax.set_xlim(0, 55)
ax.grid(axis="x", color="#e1e0d9", linewidth=0.8)
ax.grid(axis="y", visible=False)
ax.set_title("P2. Participação nas matrículas do estado (10 maiores municípios)")
ax.set_xlabel("Percentual das matrículas do estado (%)")
ax.set_ylabel("Município")
plt.show()
````

### Pergunta 3 — A infraestrutura das escolas rurais é diferente?

Os indicadores `IN_INTERNET`, `IN_BIBLIOTECA` e `IN_ENERGIA_INEXISTENTE` valem 1 (sim) ou 0 (não). Por isso, a **média** de cada indicador dentro de um grupo é a proporção de escolas que possuem o item.

````python
itens_infraestrutura = ["Internet", "Biblioteca", "Sem energia elétrica"]
indicadores = [tem_internet, tem_biblioteca, sem_energia]

percentual_urbanas = np.array([np.nanmean(indicador[urbana]) * 100 for indicador in indicadores])
percentual_rurais = np.array([np.nanmean(indicador[rural]) * 100 for indicador in indicadores])

for item, urbano, rural_pct in zip(itens_infraestrutura, percentual_urbanas, percentual_rurais):   # laço apenas para exibir
    print(f"{item:<22} urbanas: {urbano:5.1f}% | rurais: {rural_pct:5.1f}%")
print()
print(f"Escolas rurais sem internet: {np.sum(rural & (tem_internet == 0))} | sem energia elétrica: {np.sum(rural & (sem_energia == 1))}")

fig, ax = plt.subplots()
posicoes_itens = np.arange(len(itens_infraestrutura))
largura = 0.38
barras_urbanas = ax.bar(posicoes_itens - largura / 2, percentual_urbanas, width=largura, color=CORES[0], label="Escolas urbanas")
barras_rurais = ax.bar(posicoes_itens + largura / 2, percentual_rurais, width=largura, color=CORES[1], label="Escolas rurais")
for barras in (barras_urbanas, barras_rurais):
    for barra in barras:
        ax.text(barra.get_x() + barra.get_width() / 2, barra.get_height() + 1.5,
                f"{barra.get_height():.1f}%".replace(".", ","), ha="center", fontsize=10, color=COR_TEXTO_SECUNDARIO)
ax.set_xticks(posicoes_itens, itens_infraestrutura)
ax.set_ylim(0, 105)
ax.set_title("P3. Infraestrutura das escolas em atividade: urbanas × rurais")
ax.set_xlabel("Item de infraestrutura")
ax.set_ylabel("Percentual das escolas (%)")
ax.legend(loc="upper right")
plt.show()
````

| `data_pas` | data e hora (UTC) da passagem do satélite que detectou o foco | extração do ano e do mês |

| `lat` / `lon` | coordenadas do foco (graus decimais) | filtragem geográfica (sul do estado) |
| **focos por mês / por ano** | quantidade de focos, obtida por contagem | variável quantitativa principal |

Cada linha dos arquivos é **um foco de calor**, ou seja, um ponto em que o satélite detectou temperatura elevada compatível com fogo na vegetação. Um foco não equivale a uma queimada: um mesmo incêndio grande pode gerar vários focos.

### 3. Quantidade de observações

````python
url_base_queimadas = "https://dataserver-coids.inpe.br/queimadas/queimadas/focos/csv/anual/EstadosBr_sat_ref/AM/"
pasta_queimadas = os.path.join("dados", "inpe_queimadas")
anos_queimadas = np.arange(2003, 2025)
os.makedirs(pasta_queimadas, exist_ok=True)

datas_focos, municipios_focos, latitudes_focos, longitudes_focos = [], [], [], []

for ano in anos_queimadas:   # um arquivo por ano
    nome_zip = f"focos_br_am_ref_{ano}.zip"
    caminho_zip = os.path.join(pasta_queimadas, nome_zip)
    if not os.path.exists(caminho_zip):
        requisicao = urllib.request.Request(url_base_queimadas + nome_zip, headers={"User-Agent": "Mozilla/5.0"})
        with urllib.request.urlopen(requisicao, timeout=120) as resposta, open(caminho_zip, "wb") as arquivo:
            arquivo.write(resposta.read())

    with zipfile.ZipFile(caminho_zip) as arquivo_zip:
        linhas = arquivo_zip.read(arquivo_zip.namelist()[0]).decode("utf-8").splitlines()

    # Colunas: id_bdq, foco_id, lat, lon, data_pas, pais, estado, municipio, bioma
    tabela = np.genfromtxt(linhas, delimiter=",", skip_header=1, usecols=(2, 3, 4, 7),
                           dtype=str, autostrip=True, encoding="utf-8")
    datas_focos.append(tabela[:, 2])
    municipios_focos.append(tabela[:, 3])
    latitudes_focos.append(tabela[:, 0])
    longitudes_focos.append(tabela[:, 1])

datas_focos = np.concatenate(datas_focos)
municipios_focos = np.concatenate(municipios_focos)
latitudes_texto = np.concatenate(latitudes_focos)
longitudes_texto = np.concatenate(longitudes_focos)

print("Arquivos anuais:", anos_queimadas.size, f"({anos_queimadas[0]} a {anos_queimadas[-1]})")
print("Focos de calor carregados:", datas_focos.size)
print("Exemplo de registros:")
print(np.column_stack((datas_focos[:3], municipios_focos[:3], latitudes_texto[:3], longitudes_texto[:3])))
````

### 4. Valores ausentes e preparação dos dados

A partir das datas, cada foco recebeu um ano e um mês. A **contagem** de focos por mês foi feita de forma vetorizada com `np.bincount`, gerando uma matriz de **22 anos × 12 meses**. Nenhum dos 272.982 registros tem campo vazio, e **nenhum dos 264 meses ficou sem registro**, então não foi necessário tratamento de ausentes.

````python
# Ano e mês de cada foco a partir do texto "AAAA-MM-DD hh:mm:ss"
partes_data = np.char.split(np.char.partition(datas_focos, " ")[:, 0], "-")
ano_foco = np.array([partes[0] for partes in partes_data], dtype=int)
mes_foco = np.array([partes[1] for partes in partes_data], dtype=int)
latitude_foco = latitudes_texto.astype(float)

# 4. Identificação de valores ausentes
print("Campos vazios -> data:", np.sum(datas_focos == ""), "| município:", np.sum(municipios_focos == ""),
      "| latitude:", np.sum(latitudes_texto == ""), "| longitude:", np.sum(longitudes_texto == ""))

# Agregação: contagem de focos em cada mês de cada ano -> matriz 22 anos x 12 meses
indice_ano_mes = (ano_foco - anos_queimadas[0]) * 12 + (mes_foco - 1)
focos_por_mes = np.bincount(indice_ano_mes, minlength=anos_queimadas.size * 12).reshape(anos_queimadas.size, 12)
focos_por_ano = np.sum(focos_por_mes, axis=1)
serie_mensal = focos_por_mes.ravel()   # 264 meses em ordem cronológica

print(f"Meses sem nenhum registro: {np.sum(serie_mensal == 0)} de {serie_mensal.size}")
print(f"Conferência: soma da matriz = {np.sum(focos_por_mes)} focos")
print()
print("Focos por ano:")
print(np.column_stack((anos_queimadas, focos_por_ano)).T)
````

### 5 a 9. Mínimo, máximo, média, mediana, desvio padrão e quartis

As estatísticas foram calculadas para duas séries derivadas: o **total de focos por ano** (22 valores) e o **total de focos por mês** (264 valores).

````python
def estatisticas(valores):
    q1_valor, q3_valor = np.percentile(valores, [25, 75])
    return np.min(valores), np.max(valores), np.mean(valores), np.median(valores), np.std(valores), q1_valor, q3_valor

print(f"{'Série':<26} {'n':>4} {'Mínimo':>8} {'Máximo':>8} {'Média':>9} {'Mediana':>9} {'Desvio':>9} {'Q1':>8} {'Q3':>8}")
for nome, valores in [("Focos por ano", focos_por_ano), ("Focos por mês", serie_mensal)]:   # laço apenas para a tabela
    minimo, maximo, media, mediana, desvio, q1_valor, q3_valor = estatisticas(valores)
    print(f"{nome:<26} {valores.size:>4} {minimo:>8,.0f} {maximo:>8,.0f} {media:>9,.1f} {mediana:>9,.1f} "
          f"{desvio:>9,.1f} {q1_valor:>8,.1f} {q3_valor:>8,.1f}")
print()
print(f"Ano com menos focos: {anos_queimadas[np.argmin(focos_por_ano)]} | ano com mais focos: {anos_queimadas[np.argmax(focos_por_ano)]}")
linha_max, coluna_max = np.unravel_index(np.argmax(focos_por_mes), focos_por_mes.shape)
print(f"Mês com mais focos em toda a série: {abreviacoes_meses[coluna_max]}/{anos_queimadas[linha_max]} ({focos_por_mes[linha_max, coluna_max]:,} focos)")
````

### 10. Filtragens com máscaras NumPy

Máscaras booleanas selecionam os focos da estação de queimadas (agosto a outubro), os focos do sul do estado (latitude menor que −7°) e dois períodos (2003–2012 e 2020–2024). Também foi feita uma agregação por município com `np.unique` e `np.bincount`.

````python
# 10. Filtragens com máscaras NumPy
estacao_de_queimadas = (mes_foco >= 8) & (mes_foco <= 10)          # agosto a outubro
sul_do_estado = latitude_foco < -7                                   # faixa ao sul do paralelo 7° S
periodo_recente = ano_foco >= 2020
periodo_inicial = ano_foco <= 2012

print(f"Focos entre agosto e outubro: {np.mean(estacao_de_queimadas) * 100:.1f}% do total "
      f"(2003–2012: {np.mean(estacao_de_queimadas[periodo_inicial]) * 100:.1f}% | 2020–2024: {np.mean(estacao_de_queimadas[periodo_recente]) * 100:.1f}%)")
print(f"Focos ao sul da latitude 7° S: {np.mean(sul_do_estado) * 100:.1f}% do total "
      f"(2003–2012: {np.mean(sul_do_estado[periodo_inicial]) * 100:.1f}% | 2020–2024: {np.mean(sul_do_estado[periodo_recente]) * 100:.1f}%)")
print(f"Média anual 2003–2012: {np.mean(focos_por_ano[anos_queimadas <= 2012]):,.0f} focos | "
      f"2020–2024: {np.mean(focos_por_ano[anos_queimadas >= 2020]):,.0f} focos")
print()

# Agregação por município
nomes_municipios_focos, indice_municipio_foco = np.unique(municipios_focos, return_inverse=True)
focos_por_municipio = np.bincount(indice_municipio_foco)
ordem_focos = np.argsort(focos_por_municipio)[::-1]

print(f"Municípios com focos registrados: {nomes_municipios_focos.size}")
print(f"Os 5 municípios com mais focos concentram {np.sum(focos_por_municipio[ordem_focos[:5]]) / focos_por_municipio.sum() * 100:.1f}% do total")
print(f"Manaus: {focos_por_municipio[nomes_municipios_focos == 'MANAUS'][0]:,} focos "
      f"({focos_por_municipio[nomes_municipios_focos == 'MANAUS'][0] / focos_por_municipio.sum() * 100:.2f}%)")
````

### Etapa 2 — Preparando os dados

- **Valores ausentes:** nenhuma coluna tem campos vazios.

Esses valores **foram mantidos**: são preços possíveis de postos específicos, e não erros evidentes. Como a análise se baseia principalmente em **medianas** e em médias de milhares de coletas, esses poucos valores não alteram as conclusões.
- **Organização:** o preço foi convertido para `float` (trocando a vírgula decimal por ponto), a data foi separada em dia, mês e ano, e cada coleta recebeu um índice de mês de 0 (jan/2024) a 23 (dez/2025). Máscaras booleanas separam os produtos, Manaus e o ano de 2025.

````python
# --- Valores ausentes ---
for nome in colunas_precos:   # uma verificação por coluna
    print(f"Campos vazios em {nome:<12}: {np.sum(dados_precos[:, indice_precos[nome]] == '')}")

# --- Conversão de tipos ---
preco = np.char.replace(preco_texto, ",", ".").astype(float)          # "6,99" -> 6.99
partes_data_coleta = np.char.split(data_coleta, "/")                    # "dd/mm/aaaa"
dia_coleta = np.array([partes[0] for partes in partes_data_coleta], dtype=int)
mes_coleta = np.array([partes[1] for partes in partes_data_coleta], dtype=int)
ano_coleta = np.array([partes[2] for partes in partes_data_coleta], dtype=int)
indice_mes = (ano_coleta - 2024) * 12 + (mes_coleta - 1)                # 0 = jan/2024 ... 23 = dez/2025

# --- Dados inválidos ---
preco_invalido = (preco <= 0) | (preco > 15)
data_invalida = (dia_coleta < 1) | (dia_coleta > 31) | (mes_coleta < 1) | (mes_coleta > 12) | ~np.isin(ano_coleta, [2024, 2025])
print()
print(f"Preços inválidos (<= 0 ou > R$ 15): {np.sum(preco_invalido)} | datas inválidas: {np.sum(data_invalida)}")

# Valores atípicos pelo IQR, dentro de cada combinação capital × produto
atipico = np.zeros(preco.size, dtype=bool)
for uf in np.unique(uf_coleta):
    for produto in produtos:
        grupo = (uf_coleta == uf) & (produto_coleta == produto)
        q1_grupo, q3_grupo = np.percentile(preco[grupo], [25, 75])
        iqr_grupo = q3_grupo - q1_grupo
        atipico[grupo] = (preco[grupo] < q1_grupo - 1.5 * iqr_grupo) | (preco[grupo] > q3_grupo + 1.5 * iqr_grupo)
print(f"Coletas atípicas pelo IQR (por capital e produto): {np.sum(atipico)} ({np.mean(atipico) * 100:.2f}%)")
print("Maior preço de etanol:", np.max(preco[produto_coleta == "ETANOL"]),
      "em", uf_coleta[produto_coleta == "ETANOL"][np.argmax(preco[produto_coleta == "ETANOL"])])
print("Coletas de gasolina acima de R$ 8,50 por UF:", dict(zip(*[x.tolist() for x in np.unique(uf_coleta[(produto_coleta == "GASOLINA") & (preco > 8.5)], return_counts=True)])))

# --- Seleção e organização em arrays ---
gasolina = produto_coleta == "GASOLINA"
etanol = produto_coleta == "ETANOL"
diesel = produto_coleta == "DIESEL S10"
manaus = uf_coleta == "AM"
ano_2025 = ano_coleta == 2025

siglas_capitais = np.unique(uf_coleta)
nomes_capitais = {
    "AC": "Rio Branco", "AL": "Maceió", "AM": "Manaus", "AP": "Macapá", "BA": "Salvador", "CE": "Fortaleza",
    "DF": "Brasília", "ES": "Vitória", "GO": "Goiânia", "MA": "São Luís", "MG": "Belo Horizonte",
    "MS": "Campo Grande", "MT": "Cuiabá", "PA": "Belém", "PB": "João Pessoa", "PE": "Recife", "PI": "Teresina",
    "PR": "Curitiba", "RJ": "Rio de Janeiro", "RN": "Natal", "RO": "Porto Velho", "RR": "Boa Vista",
    "RS": "Porto Alegre", "SC": "Florianópolis", "SE": "Aracaju", "SP": "São Paulo", "TO": "Palmas",
}
rotulos_meses = [f"{abreviacoes_meses[m % 12].lower()}/{24 + m // 12}" for m in range(24)]

print()
print(f"Coletas em Manaus: {np.sum(manaus)} | gasolina: {np.sum(manaus & gasolina)} | "
      f"etanol: {np.sum(manaus & etanol)} | diesel S10: {np.sum(manaus & diesel)}")
````

### Etapa 3 — Estatísticas

Estatísticas dos preços de cada produto no conjunto das 27 capitais e em Manaus. Além das exigidas, foi incluído o **coeficiente de variação (CV = desvio padrão ÷ média)**, que indica a dispersão relativa dos preços.

````python
def resumo_precos(valores):
    q1_valor, q3_valor = np.percentile(valores, [25, 75])
    return (valores.size, np.mean(valores), np.median(valores), np.std(valores), np.min(valores),
            np.max(valores), q1_valor, q3_valor, np.max(valores) - np.min(valores))

print(f"{'Recorte':<30} {'n':>6} {'Média':>6} {'Mediana':>7} {'Desvio':>6} {'Mín':>5} {'Máx':>5} {'Q1':>5} {'Q3':>5} {'Amplit.':>7} {'CV':>6}")
for rotulo, mascara_produto in [("Gasolina", gasolina), ("Etanol", etanol), ("Diesel S10", diesel)]:   # laço apenas para a tabela
    for local, mascara_local in [("27 capitais", np.ones(preco.size, dtype=bool)), ("Manaus", manaus)]:
        n, media, mediana, desvio, minimo, maximo, q1_valor, q3_valor, amplitude = resumo_precos(preco[mascara_produto & mascara_local])
        print(f"{rotulo + ' — ' + local:<30} {n:>6} {media:>6.2f} {mediana:>7.2f} {desvio:>6.2f} {minimo:>5.2f} "
              f"{maximo:>5.2f} {q1_valor:>5.2f} {q3_valor:>5.2f} {amplitude:>7.2f} {desvio / media * 100:>5.1f}%")
````

#### Pergunta 2 — Como os preços evoluíram em Manaus em comparação com a média das capitais?

O preço médio de cada mês foi calculado com `np.bincount`: somando os preços de cada mês (usando os preços como pesos) e dividindo pelo número de coletas do mês.

````python
# Preço médio mensal: soma dos preços / número de coletas em cada mês (agregação com np.bincount)
def media_mensal(mascara):
    soma = np.bincount(indice_mes[mascara], weights=preco[mascara], minlength=24)
    quantidade = np.bincount(indice_mes[mascara], minlength=24)
    return soma / quantidade

produtos_analise = [("Gasolina", gasolina), ("Etanol", etanol), ("Diesel S10", diesel)]
media_manaus = {nome: media_mensal(mascara & manaus) for nome, mascara in produtos_analise}
media_capitais = {nome: media_mensal(mascara) for nome, mascara in produtos_analise}

for nome, _ in produtos_analise:   # laço apenas para exibir
    diferenca = media_manaus[nome] - media_capitais[nome]
    print(f"{nome:<10} Manaus: jan/24 R$ {media_manaus[nome][0]:.2f} -> dez/25 R$ {media_manaus[nome][-1]:.2f} "
          f"({(media_manaus[nome][-1] / media_manaus[nome][0] - 1) * 100:+.1f}%) | "
          f"capitais: {(media_capitais[nome][-1] / media_capitais[nome][0] - 1) * 100:+.1f}% | "
          f"diferença Manaus − capitais: jan/24 {diferenca[0]:+.2f}, dez/25 {diferenca[-1]:+.2f}, máxima {np.max(diferenca):+.2f} ({rotulos_meses[np.argmax(diferenca)]})")

fig, eixos = plt.subplots(1, 3, figsize=(15, 4.8))
for ax, (nome, _) in zip(eixos, produtos_analise):
    ax.plot(np.arange(24), media_manaus[nome], color=CORES[0], marker="o", markersize=4, label="Manaus")
    ax.plot(np.arange(24), media_capitais[nome], color=COR_REFERENCIA, linestyle="--", label="Média das 27 capitais")
    ax.set_title(nome, fontsize=12)
    ax.set_xticks(np.arange(0, 24, 6), rotulos_meses[::6])
    ax.set_xlabel("Mês da coleta")
    ax.set_ylabel("Preço médio (R$ por litro)")
eixos[0].legend(loc="upper left")
fig.suptitle("P2. Evolução do preço médio mensal: Manaus × média das capitais", x=0.01, ha="left",
             fontsize=14, fontweight="bold")
fig.tight_layout()
plt.show()
````

**Resposta:** os preços subiram em todo o país, mas **subiram mais em Manaus**, e a diferença para as demais capitais aumentou:

- **Gasolina:** em Manaus passou de R\$ 5,63 (jan/2024) para R\$ 6,99 (dez/2025), alta de **24,0%**, contra 11,7% na média das capitais. A diferença saiu de apenas **+R\$ 0,07** para **+R\$ 0,77** por litro, chegando a +R\$ 0,91 em março de 2025.
- **Etanol:** alta de **32,1%** em Manaus contra 24,3% nas capitais, e a diferença dobrou, de +R\$ 0,46 para +R\$ 0,90.
- **Diesel S10:** variou menos (+5,6% em Manaus e +1,5% nas capitais), mas também teve a diferença ampliada, de +R\$ 0,15 para +R\$ 0,40.

Também chama atenção a **estabilidade** dos preços em Manaus no fim do período: de agosto a dezembro de 2025 a gasolina ficou em R\$ 6,99, e de março a dezembro de 2025 o etanol ficou entre R\$ 5,49 e R\$ 5,50, enquanto a média das capitais continuou oscilando.

#### Pergunta 3 — Em Manaus, compensou abastecer com etanol?

Como o etanol rende menos por litro que a gasolina, a regra prática usada no Brasil é que ele compensa quando custa **até 70%** do preço da gasolina.

````python
# Regra prática: o etanol compensa quando custa até 70% do preço da gasolina
razao_etanol_gasolina = media_manaus["Etanol"] / media_manaus["Gasolina"]
compensa_etanol = razao_etanol_gasolina <= 0.70

print("Razão etanol/gasolina em Manaus:", np.round(razao_etanol_gasolina, 3))
print(f"Meses em que o etanol compensou: {np.sum(compensa_etanol)} de 24 -> {np.array(rotulos_meses)[compensa_etanol]}")
print(f"Razão média: 2024 = {np.mean(razao_etanol_gasolina[:12]):.3f} | 2025 = {np.mean(razao_etanol_gasolina[12:]):.3f}")

razao_por_capital = np.array([np.median(preco[etanol & (uf_coleta == uf)]) / np.median(preco[gasolina & (uf_coleta == uf)])
                              for uf in siglas_capitais])
print("Capitais onde o etanol compensou no período (razão das medianas <= 0,70):",
      [nomes_capitais[uf] for uf in siglas_capitais[razao_por_capital <= 0.70]])

fig, ax = plt.subplots()
ax.plot(np.arange(24), razao_etanol_gasolina, color=CORES[0], marker="o", label="Manaus: etanol ÷ gasolina")
ax.axhline(0.70, color=CORES[1], linewidth=2, linestyle="--", label="Limite de 70%")
ax.fill_between(np.arange(24), 0.60, 0.70, color="#f0efec", zorder=0)
ax.text(23, 0.655, "abaixo de 70%: etanol compensa", ha="right", fontsize=10, color=COR_TEXTO_SECUNDARIO)
ax.set_title("P3. Relação entre o preço do etanol e o da gasolina em Manaus")
ax.set_xlabel("Mês da coleta")
ax.set_ylabel("Preço do etanol ÷ preço da gasolina")
ax.set_xticks(np.arange(0, 24, 3), rotulos_meses[::3])
ax.set_ylim(0.60, 0.82)
ax.legend(loc="upper left")
plt.show()
````

Em 2025, a razão média subiu para **0,77**, porque o etanol encareceu proporcionalmente mais que a gasolina, e ficou acima do limite em todos os meses. Considerando as medianas de todo o período, o etanol só compensou em quatro capitais: Brasília, Campo Grande, Cuiabá e São Paulo, todas do Centro-Oeste e do Sudeste.

8. **O diesel S10 foi o combustível mais estável:** em Manaus variou 5,6% em dois anos e teve o menor coeficiente de variação entre os três produtos (3,9%), embora também tenha ficado mais caro que a média das capitais durante todo o período.
9. **Preços caros tendem a andar juntos:** capitais com gasolina mais cara tendem a ter diesel (r = 0,73) e etanol (r = 0,60) mais caros, mas há exceções importantes, como Macapá, que combina gasolina barata (R\$ 5,94) com o etanol mais caro do país (R\$ 5,33).

*Limites da análise:* os dados mostram diferenças de preço, mas não permitem identificar suas causas, como custos de transporte, impostos ou concorrência. Além disso, a análise se restringe às capitais e aos postos pesquisados pela ANP.
